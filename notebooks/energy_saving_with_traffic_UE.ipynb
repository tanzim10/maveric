```python
import pandas as pd
import numpy as np
import json

from radp.digital_twin.utils.traffic_demand_simulation import TrafficDemandModel
```

# Load Data

```python
topology = pd.read_csv("data/traffic_data/topology.csv")
spatial_params = json.load(open("data/traffic_data/spatial_params.json"))
time_params = json.load(open("data/traffic_data/time_params.json"))
num_ues_per_tick = 200

print(f"\n Target UEs per tick: {num_ues_per_tick}")
print(f"\n Topology: {topology.head(3)}")
print(f"\n Spatial Parameters {json.dumps(spatial_params, indent=2)}")
print(f"\n Time Parameters {json.dumps(time_params, indent=2)}")
```

# Run `TrafficDemandModel` to create new UE Data 

```python
traffic_demand = TrafficDemandModel()

ue_data_per_tick, spatial_layout = traffic_demand.generate_traffic_demand(topology, spatial_params, time_params, num_ues_per_tick)

```

```python
for df_no, (tick, df) in enumerate(ue_data_per_tick.items()):
    if df_no >= 3:
        break
    print(f"Tick: {tick}")
    print(df.head(4))
    print("-" * 40)
```
